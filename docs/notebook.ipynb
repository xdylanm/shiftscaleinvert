# Attenuverter Transfer Functions

```python
import matplotlib.pyplot as plt
import numpy as np

alpha = np.linspace(0,1,51)
```

## Basic

$H(\alpha)=2\alpha - 1$

```python
H_b = 2*alpha - 1

plt.plot(alpha,H_b)
plt.xlabel("alpha")
plt.ylabel("H")
plt.title("Basic")
plt.grid(True)
plt.xticks(np.linspace(0,1,11))
plt.show()
```

## Balanced

$H(\alpha) = G\frac{(2\alpha -1)(3+3G-2\alpha)}{(3-2\alpha)(1+G+2G\alpha)}$

```python
N = 4
G_vals = np.linspace(0.5,2,N)
def H_bal(al, G):
    return G * (2*al - 1)*(3 + 3*G - 2*al)/(3 - 2*al)/(1 + G + 2*G*al)

for i,G in enumerate(G_vals):
    plt.plot(alpha, H_bal(alpha, G) / G)

plt.xlabel("alpha")
plt.ylabel("H")
plt.title("Balanced (normalized to G)")
plt.grid(True)
plt.legend(["G=0.5","G=1.0","G=1.5","G=1.2"])
plt.xticks(np.linspace(0,1,11))
plt.show()
```

## Precision

$$\begin{align*}
H(\alpha) &= 2A(\alpha)-1\\
A(\alpha) &= \alpha\frac{1-\alpha + R_s/R_g}{2\alpha(1 - \alpha)  + R_s/R_g}
\end{align*}$$

```python
N = 4
r_vals = [0.1, 0.22, 0.33, 0.47, 0.68, 1.0]
def H_prec(al, r):
    A = al * (1 - al + r)/(2*al*(1 -al) + r)
    return 2*A - 1

for i,r in enumerate(r_vals):
    plt.plot(alpha, H_prec(alpha, r))

plt.xlabel("alpha")
plt.ylabel("H")
plt.title("Precision (r = Rs/Rg)")
plt.grid(True)
plt.legend(["r=0.1","r=0.22","r=0.33","r=0.47","r=0.68","r=1.00"])
plt.xticks(np.linspace(0,1,11))
plt.show()
```
